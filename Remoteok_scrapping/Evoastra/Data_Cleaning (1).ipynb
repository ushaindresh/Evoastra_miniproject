```python
import pandas as pd
```

```python
df = pd.read_csv("/content/remoteok_jobs_day4 (1).csv")
```

```python
print("Initial shape:", df.shape)
```

```python
# Standardize missing values

df.replace(
    ["N/A", "n/a", "NA", "", "None", "null", "NULL"],
    pd.NA,
    inplace=True
)
```

```python
# Remove duplicate entries

df.drop_duplicates()
```

```python
# Handle missing values ()

critical_columns = ["title", "company", "url"]

existing_critical_cols = [col for col in critical_columns if col in df.columns]
df = df.dropna(subset=existing_critical_cols)
```

```python
# Standardize text columns (lower case and strip white spaces)

text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.lower().str.strip()
```

```python
# Save clean Dataset

output_file = "cleaned Dataset.csv"
df.to_csv(output_file, index=False)

print(f"Cleaned data saved as: {output_file}")
```

```python
# Print summary statistics

print("\n===== SUMMARY STATISTICS =====")
print("Total jobs after cleaning:", len(df))
print("\nMissing values per column:")
print(df.isna().sum())

print("\nFinal shape:", df.shape)
```
